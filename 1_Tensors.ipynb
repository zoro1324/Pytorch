```python
import torch
```

```python
def eval(tensor):
    print("="*50)
    print("Shape:",tensor.shape)
    print("Dimension:",tensor.dim())
    print("Tensor:\n",tensor)
    print("="*50)
```

```python
tensor  = torch.tensor(
    [1,2,3], 
)
eval(tensor=tensor)
```

```python
tensor = torch.tensor([
    [
        [1,2,3],
        [4,5,8]
    ],
    [
        [1,25,3],
        [45,5,85]
    ],
    [
        [11,2,38.0],
        [43,5,81],
    ]
]
)

eval(tensor=tensor)
```

```python
tensor = torch.rand(size=(1200,1200,3),dtype=float)
eval(tensor=tensor)
```
