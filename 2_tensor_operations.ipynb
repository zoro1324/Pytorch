```python
import torch 

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device
```

```python
tensor_1 = torch.randn(
    size=(3,3),
    device=device,
    requires_grad=False
)

tensor_2 = torch.rand_like(
    input=tensor_1
)

print(f"Tensor 1 : {tensor_1} \nTensor 2 : {tensor_2}")
```

```python
%%time
torch.matmul(tensor_1,tensor_2)
```
